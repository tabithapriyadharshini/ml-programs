```python
import numpy as np
from scipy import sparse
matrix = np.array([
    [0, 0],
    [0, 1],
    [3, 0]
])
matrix_sparse = sparse.csr_matrix(matrix)
print("Original Matrix:")
print(matrix)
print("\nSparse Matrix:")
print(matrix_sparse)
print("\nSparse Matrix as Array:")
print(matrix_sparse.toarray())
```
