```python
import pandas as pd

# Create dataset
data = {
    "Name": ["Sukumar", "Rahul", "Kiran", "Arjun"],
    "Age": [22, 23, 21, 24],
    "Marks": [85, 78, 92, 65]
}

df = pd.DataFrame(data)

# Rename columns
df = df.rename(columns={
    "Name": "Student_Name",
    "Age": "Student_Age",
    "Marks": "Student_Marks"
})

# Display the data
print(df)
```
