```python
import pandas as pd
import random

# Number of students
n = 20

# Create simulated data
data = {
    "Student_ID": range(1, n + 1),
    "Age": [random.randint(18, 25) for _ in range(n)],
    "Study_Hours": [round(random.uniform(1, 8), 1) for _ in range(n)],
    "Attendance": [random.randint(60, 100) for _ in range(n)],
    "Marks": [random.randint(40, 100) for _ in range(n)]
}

# Create DataFrame
df = pd.DataFrame(data)

# Display dataset
print(df)

# Save dataset as CSV file
df.to_csv("student_dataset.csv", index=False)

print("\nDataset saved as student_dataset.csv")
```
